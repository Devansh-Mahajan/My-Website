```python
import numpy as np
import matplotlib.pyplot as pl
```

```python
def Vec (x, y):
    v = np.array ([[x], [y]])
    return v.astype (float)
```

```python
def PlotVec (p, u, c):
    pl.arrow (p [0] [0], p [1] [0], u [0] [0], u [1] [0],
        color = c,
        head_width = 0.5,
        length_includes_head = True,
        fill = False)
```

```python
a = Vec (-7, -3)
a
```

```python
b = Vec (13, -5)
b
```

```python
c = Vec (2, 11)
c
```

```python
v = b - a
v
```

```python
w = c - a
w
```

```python
o = Vec (0, 0)
```

```python
PlotVec (a, v, 'blue')
PlotVec (a, w, 'red')
pl.plot (0, 0, 'o', color = 'black')
PlotVec (o, v, '#8888FF')
PlotVec (o, w, '#FF8888')
pl.gca().set_aspect('equal')
pl.grid (True)
```
