```python
import numpy as np
import numpy.linalg as la
```

1. Gerade

Ein fester Punkt auf der Geraden

```python
p = np.array ([
    [-3],
    [ 2]
])
```

Richtungsvektor

```python
u = np.array ([
    [2],
    [2]
])
```

2. Gerade

Ein fester Punkt auf der Geraden

```python
q = np.array ([
    [4],
    [3]
])
```

Richtungsvektor

```python
v = np.array ([
    [ 1],
    [-2]
])
```

Gleichungssystem aufstellen und lösen

```python
A = np.hstack ((u, -v))
A
```

```python
b = q - p
b
```

```python
x = la.solve (A, b)
x
```

Probe

```python
[[s], [t]] = x
print (s, t)
```

```python
p + s * u
```

```python
q + t * v
```
