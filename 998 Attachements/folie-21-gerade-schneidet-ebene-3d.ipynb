```python
import numpy as np
import numpy.linalg as la
```

```python
def Vec (x, y, z):
    v = np.array ([[x], [y], [z]])
    return v.astype (float)
```

Gerade

Ein fester Punkt auf der Geraden

```python
p = Vec (1, 0, 2)
p
```

Richtungsvektor

```python
u = Vec (1, 2, 1)
u
```

Ebene

Ein fester Punkt auf der Ebene

```python
q = Vec (2, 3, 1)
q
```

Erster Richtungsvektor

```python
v = np.array ([
    [4],
    [2],
    [9]
])
```

Zweiter Richtungsvektor

```python
w = np.array ([
    [3],
    [1],
    [6]
])
```

Gleichungssystem aufstellen und lösen

```python
A = np.hstack ((u, -v, -w))
A
```

```python
b = q - p
b
```

```python
x = la.solve (A, b)
x
```

Probe

```python
[[r], [s], [t]] = x
print (r, s, t)
```

```python
p + r * u
```

```python
q + s * v + t * w
```
